```python
# Getter Method used to access private attr. outside the class

class person:
  def __init__(self,name,job):
    self.name = name
    self.__job = job

  def getter(self):
    print("The job is",self.__job)

ob = person("Rishiraj","Data Analyst")
print(ob.name)
ob.getter()
```

```python
# Setter
# Getter

class person:
  def __init__(self,name,job):
    self.name = name
    self.__job = job

  def getter(self):
    print("The job is",self.__job)

  def setter(self,newjob):
    self.__job = newjob


ob = person("Rishiraj","Data analyst")
print(ob.name)
ob.getter()
ob.setter("Data Scientist")
ob.getter()
```

```python
class Employee: # Parent class
  def __init__(self,name,empid):
    self.name = name
    self.empid = empid

  def display(self):
    print("The name of the employee is: ",self.name,"with emp id: ",self.empid)


class tech(Employee): # child class
  def __init__(self,name ,empid,team):
    #super().__init__(name,empid)
    Employee.__init__(self,name,empid) # Direct way
    self.team = team

  def display(self):
    print(self.name,self.empid,self.team)

ob = tech("Rohit",5015,"Python")
ob.display()
```

```python
# Parent class 1
class Father:
    def father_info(self):
        print("Father: He is a teacher.")

# Parent class 2
class Mother:
    def mother_info(self):
        print("Mother: She is a doctor.")

# Child class inheriting from both parents
class Child(Father, Mother):
    def child_info(self):
        print("Child: I am a student.")

# Creating object of Child class
c = Child()

# Calling methods from all classes
c.father_info()
c.mother_info()
c.child_info()
```

```python
class bank_account:
    def __init__(self,name,bank_balance,account_number):
        self.name=name
        self.__bank_balance=bank_balance
        self.__account_number=account_number

    def deposit(self,Amount):  #getter
        print("*** By Using Getter Method *** ")
        print("Before Deposit Amount Is :- ",self.__bank_balance)
        self.__bank_balance += Amount
        print("After Deposit Amount Is :- ",self.__bank_balance)

    def withdraw(self,amount):   # setter
        print("*** By Using Setter Method *** ")
        print("The Withdraw Amount is : ",amount)
        self.__bank_balance -= amount

    def display(self):
        print(" * * *   Bank Details    * * *    ")
        print("Account Holder Name :- ",self.name)
        print("Account No  :- ",self.__account_number)
        print("Available Balance :- ",self.__bank_balance)


ob1= bank_account("Vaibhav Swami",500,43678947889)

ob1.deposit(5000)
print()
ob1.withdraw(1000)
print()
ob1.display()
```

```python
from abc import ABC , abstractmethod
# abc - Abstract base class module
```

```python
# Both ABC and abstractmethod are mandatory to achieve abstraction
```

```python
class Vehicle(ABC):  # Abstract class(Parent class)
  @abstractmethod
  def start(self):
    pass

  @abstractmethod
  def stop(self):
    pass

  def display(self):
    print("Hello")
```

```python
class Car(Vehicle):   # Concrete class(Child class)
  def start(self):
    print("The Car started")

  def stop(self):
    print("The Car stopped")

ob = Car()
ob.start()
```

```python
# Design a ride booking system with an abstract class RideBooking having an abstarct method named calculate_fare()
# Create tweo concrete classes BikeRide and  CarRide implementing this method differently
# Try to achieve abstraction
```

```python
from abc import ABC , abstractmethod
```

```python
class RideBooking(ABC):
  @abstractmethod
  def calculate_fare(self,distance):
    pass

class BikeRide(RideBooking):
  def calculate_fare(self, distance):
    print("Calcualted Fair is: ", distance*10)

class CarRide(RideBooking):
  def calculate_fare(self, distance):
    print("Calcualted Fair is: ", distance*20)


ob1 = BikeRide()
ob2 = CarRide()
n= int(input())
ob1.calculate_fare(n)
ob2.calculate_fare(n)
```

```python
class Student:
    def __init__(self, name, rollno, marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks


    def grade(self):
        if self.marks >= 80:
            return "Grade A"
        elif self.marks >= 60:
            return "Grade B"
        elif self.marks >= 40:
            return "Grade C"
        else:
            return "Grade D"



s1 = Student("Rahul", 101, 85)
s2 = Student("Anita", 102, 62)
s3 = Student("Vansh", 103, 38)


print(s1.name, "-", s1.grade())
print(s2.name, "-", s2.grade())
print(s3.name, "-", s3.grade())
```

```python
class teacher:
    def __init__(self, name, subject):
        self.name = name
        self.subject = subject


    def teach(self):
        print("Teacher name:", self.name)
        print("Subject:", self.subject)



class mathsteacher(teacher):
    def __init__(self, name, subject):
        super().__init__(name, subject)



class scienceteacher(teacher):
    def __init__(self, name, subject):
        super().__init__(name, subject)




m1 = mathsteacher("Meghana", "Mathematics")
s1 = scienceteacher("Laasya", "Science")


m1.teach()
print()
s1.teach()
```

```python
# Create an Online Ordering System

# Create an abstract class named Order with process_order() as an abstract method inside it

# Create two subclasses named OnlineOrder and StoreOrder

# Your task here is to implement this entire structure and show the use of abstraction

# Time till 10:09 pm
```

```python

from abc import ABC ,abstractmethod

class order(ABC):
    @abstractmethod
    def processes_order(self):
        print("this is from abstract method")

    def display(self):
        print("this is from abstract class")

class online_order(order):
    def processes_order(self):
        print("this is from online_order method")

class store_order(order):
    def processes_order(self):
        print("this is from store_order method")

ob=store_order()
ob.processes_order()
ob1=online_order()
ob1.processes_order()
ob1.display()
```
