**Rules of naming a variable**

1) A variable name can start with an alphabet  or an underscore( _ )

2) A variable can't start with an number but it can contain a number.

3) Can't have special characters( !,@,#,$,%,^,&,etc)

4) Can't have space in a variable

5) Can't use keywords(reserved words - which perfrom some tasks) as variable name. Eg: for, while, if else elif, in , not in, etc

6) Variables are case sensitive

```python
# int-> float
a=10
b=-50
c=0
print(float(a))
print("Original=",a)
```

```python
d=float(b)
print(d)

c=float(c) # c=float(0)
print(c)
```

```python
# float -> int
a=10.7
b=-100.5
d=0.999999
print(int(a))
print(int(b))
print(int(c))
```

```python
# int, float -> str

a=10
b=-50
c=100.789

print(a+b)
print(str(a)) # "10"
print(str(b)) # "-50"
a=str(a)  # 10 -> "10"
c=str(c)  # 100.789 -> "100.789"
print(a+c)
```
