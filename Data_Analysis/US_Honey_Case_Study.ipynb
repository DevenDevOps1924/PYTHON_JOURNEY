# Pattern analysis with visualisation

## 1. Check the relation between number of colonies and production

```python
sns.scatterplot(data=df,x="colonies_number",y="production")
plt.title("Colonies no VS Producton ")
plt.xlabel("colonies_number")
plt.ylabel("production")
plt.show()
```

## 2. Check the relation between production and average price

```python
sns.scatterplot(x="production",y="average_price",data=df)
plt.title("production no VS average_price ")
plt.xlabel("production")
plt.ylabel("average_price")
plt.show()
```

## Q3. Which state is having the highest honey production through out all these years? represent it using a bar graph

##Q4. Which state is having the highest and the lowest yield per colony? use bar graph

```python
df.groupby('state')['yield_per_colony'].mean().sort_values(ascending = False)
```

```python
warnings.filterwarnings('ignore')
sns.barplot(x='yield_per_colony',y='state' ,data=df,palette='hsv')
plt.title("state VS yield_per_colony ")
plt.xlabel("state")
plt.ylabel("yield_per,colony")
plt.show()
```
