```python
sales = [500, 450, 700, 350, 450, 800, 950]
days = ['M','T','W','Th','F','Sa','S']

plt.plot(days, sales)
plt.show()
```

```python
sales = [500, 450, 700, 350, 450, 800, 950]
days = ['M','T','W','Th','F','Sa','S']

plt.plot(days, sales, marker = 'o')
plt.xlabel("Days")
plt.ylabel("Ice-Cream Sales")
plt.title("A week's ice-cream sales 'Line PLot'")
plt.grid(True)
plt.show()
```

```python
profit = [500, 450, 700, 350, 450, 800, 950]
years = [1,2,3,4,5,6,7]

plt.figure(figsize=(6,3))
plt.plot(years, profit, marker = 'o', color = 'green', markerfacecolor='yellow', linestyle="-.", linewidth = 7)
plt.xlabel("Year")
plt.ylabel("Profit values")
plt.title("Profit over the years")
plt.grid(True)
plt.show()

# xticks, yticks - range
```

# Pie Plot - comparing proportion then we use pie chart

```python
fruits = ['apple', 'mango',' kiwi']
liked_percentage=[35, 40, 25]

plt.pie(liked_percentage, labels = fruits)
plt.show()
```

```python
fruits = ['apple', 'mango',' kiwi']
liked_percentage=[35, 40, 25]

plt.pie(liked_percentage, labels = fruits, autopct= "%0.1f%%")
plt.show()
```

```python
sources = ["insta ads", 'billboards', 'friends', 'school', 'workplace', 'newspaper']
visitors = [30, 20, 30, 20, 15, 5]

plt.pie(visitors, labels = sources, autopct = '%0.2f%%', explode = [0,0,0,0,0,0.2])
plt.show()
```

```python
'''
1. Create scatter plots for the following data, make sure all the plots appear in the same
plane with x-labels and y-labels evenly spaced.

a. Create scatter plots with between (mpg-hp, mpg-weight), and (hp-mpg,
hp-time-to-60) with respect to each brand.
'''

# 10:55

fig = plt.subplots(2,2, figsize = (20, 10))

plt.subplot(2,2,1)
sns.scatterplot(x=df['mpg'], y= df['hp'], data = df, hue = df['brand'])

plt.subplot(2,2,2)
sns.scatterplot(x=df['mpg'], y= df['weightlbs'], data = df, hue = df['brand'])

plt.subplot(2,2,3)
sns.scatterplot(x=df['hp'], y= df['mpg'], data = df, hue = df['brand'])

plt.subplot(2,2,4)
sns.scatterplot(x=df['hp'], y= df['time-to-60'], data = df, hue = df['brand'])

plt.show()
```

```python
'''
2. Create bar plots for the following, make sure all the plots appear in the same plane with
x-labels and y-labels evenly spaced.
a. Create a bar plot that shows the visual representation of hp, mpg, weight,
time-to-60 with respect to the number of cylinders for each of the brands.

hp vs cylinder
mpg vs cylinder
weightlbs vs cylinder
time-to-60 vs cylinder
'''

# 11:14

fig = plt.subplots(2,2, figsize = (20, 10))

plt.subplot(2,2,1)
sns.barplot(x=df['cylinders'], y= df['hp'], data = df, hue = df['brand'])

plt.subplot(2,2,2)
sns.barplot(x=df['cylinders'], y= df['mpg'], data = df, hue = df['brand'])

plt.subplot(2,2,3)
sns.barplot(x=df['cylinders'], y= df['weightlbs'], data = df, hue = df['brand'])

plt.subplot(2,2,4)
sns.barplot(x=df['cylinders'], y= df['time-to-60'], data = df, hue = df['brand'])

plt.show()
```

```python
'''
b. Create a bar plot that shows the visual representation of hp, mpg, weight and
time-to-60 with respect to the year, for each brand.
'''

# --11:35

fig = plt.subplots(2,2, figsize = (20, 10))

plt.subplot(2,2,1)
sns.barplot(x=df['year'], y= df['hp'], data = df, hue = df['brand'])

plt.subplot(2,2,2)
sns.barplot(x=df['year'], y= df['mpg'], data = df, hue = df['brand'])

plt.subplot(2,2,3)
sns.barplot(x=df['year'], y= df['weightlbs'], data = df, hue = df['brand'])

plt.subplot(2,2,4)
sns.barplot(x=df['year'], y= df['time-to-60'], data = df, hue = df['brand'])

plt.show()
```

```python
'''
3. Create box plots for the following:
a. Create box plots for the columns in the step1 with respect to the number of
cylinders and brand respectively.


hp vs cylinder
mpg vs cylinder
weightlbs vs cylinder
time-to-60 vs cylinder
'''

# 11:45

plt.figure(figsize=(12, 10))
plt.subplot(2, 2, 1)
sns.boxplot(x='cylinders', y='hp', hue='brand', data=df)
plt.title('HP vs Cylinders')
plt.subplot(2, 2, 2)
sns.boxplot(x='cylinders', y='mpg', hue='brand', data=df)
plt.title('MPG vs Cylinders')
plt.subplot(2, 2, 3)
sns.boxplot(x='cylinders', y='weightlbs', hue='brand', data=df)
plt.title('Weight vs Cylinders')
plt.subplot(2, 2, 4)
sns.boxplot(x='cylinders', y='time-to-60', hue='brand', data=df)
plt.title('Time-to-60 vs Cylinders')
plt.tight_layout()
plt.show()
```
