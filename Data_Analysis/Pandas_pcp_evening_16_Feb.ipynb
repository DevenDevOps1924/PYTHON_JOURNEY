# Group By

**Divide the data into multiple groups** based on some category.
And then perfrom analysis on the formed groups.


Basically, group similar type of data(similiar based on some category-> gender, department, class, survived etc). And perfrom **aggregation operations**( sum, max, min, avg, count, mode, medium, etc)

```python
df
```

```python
df.dtypes
```

```python
for i in df.columns:
  if df[i].dtype=="object" or df[i].dtype=="category":
    m=df[i].mode()
    df.fillna({i:m[0]},inplace=True)
  else:
    m=df[i].mean()
    df.fillna({i:m},inplace=True)

# Homework-> try to understand this

# Replace the null value in object or sting col with the mode.
#  And in numerical col with the mean.
```

```python
df
```

```python
 #  2 grps based on sex->   Male      Female
 #                         -----------------
 #            Avg age ->      x          y
 #            Avg fare->      x          y
 #
```

```python
# Group the data based on sex and find avg age of the groups

print(df.groupby("sex")["age"].mean())
```

```python
# Group data based on class, find the avg fare of the groups
print(df.groupby("class")["fare"].mean())
```

```python
# Avg age , based on alive or not
print(df.groupby("alive")["age"].mean())
```

```python
# Avg age based on the alive and sex
print(df.groupby(["sex","alive"])["age"].mean())
```
