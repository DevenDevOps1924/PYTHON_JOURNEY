```python
# Handling null values

# fill them -- specific value, a calculated value, mean, median, mode, ffill, bfill
# drop them
```

```python
df = df.dropna(subset = ['product'])
```

```python
# drops the row which is having null value in product column
```

```python
df['brand'] = df['brand'].fillna("Unknown")
```

```python
# replace the null values in description col with any value of your choice

df['description'] = df['description'].fillna("Description not available")
```

```python
df.isnull().sum()
```
