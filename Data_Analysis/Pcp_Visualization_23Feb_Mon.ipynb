Adjecent Bars

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
import pandas as pd
```

```python
subjects=["Maths","Science","English","SST"]
girls=[87,91,89,82]
boys=[80,93,70,60]
x=np.arange(1,len(subjects)+1) #[1 2 3 4]



# x-> represents the positions of the bars on the x-axis.
# It is just numerical locations where bars will be drawn.

# plt.bar(x_position,height,width)
#           x   ,  marks , width of bars

# Position- of bar on x axis
# height-> values or the height of the bar (marks)
# width=wdith of bars you want

width=0.4
plt.bar(x-width/2,girls,width=0.4,label="Girls")
# x      - width/2 ->
#[1,2,3,4] - 0.2 = [0.8,1.8,2.8,3.8]
# It shifts the bar positions slightly to the LEFT of the center.

# Why divide by 2?
# - Each bar has a width, if If width = 0.4:
#   Each bar occupies 0.4 space.
#   To place two bars side by side symmetrically:
#     -First bar → shift left by 0.2
#     -Second bar → shift right by 0.2

plt.bar(x + width/2, boys, width=width,label="Boys")

# Later we replace numbers with names using:
# xticks-> the values being shown on the x axis
# plt.xticks(positions, labels)-> used to change the x labels or xticks
plt.xticks(x, subjects) # instead of [1,2,3,4] show the subjects
plt.legend()
plt.show()
```

# Pie Chart

Is a circular chard divided into slices, where each slice represents a portion(percentage) of the total. It's used to compare the contribution of different categories in the total.

And each slice shows how much is a category contributing to the total.

Used to compare parts of a while.

Ex-> sales of each category of product, number of people in each department, Survey results of population, etc etc...

```python
import matplotlib.pyplot as plt
```

```python
activity=["Rent","Travel","Food","Others","Grocery"]
expense=[8000,2000,6000,1000,3000]
# plt.pie(values,labels,customizations.....)

plt.title("Monthly Expenses")
plt.pie(expense,labels=activity)
plt.show()
```

Show percentage->


- 2 decimal places in pecentage = 1.2f%
autopct= "%1.2f%%"

- 3 deciaml places in percentage = 1.3f%
autopct= "% 1.3f% %"

- 1 decimal = %1.1f%%

```python
#                 width,height
plt.figure(figsize=(5,5))
activity=["Rent","Travel","Food","Others","Grocery"]
expense=[8000,2000,6000,1000,3000]
# plt.pie(values,labels,customizations.....)

plt.title("Monthly Expenses")
#                        2 decimal places= 1.2f% , 1.3f%, 1.4f%
plt.pie(expense,labels=activity,autopct="%1.2f%%")
plt.show()
```

```python
#                 width,height
plt.figure(figsize=(5,5))
activity=["Rent","Travel","Food","Others","Grocery"]
expense=[8000,2000,6000,1000,3000]
# plt.pie(values,labels,customizations.....)

plt.title("Monthly Expenses")
#                        2 decimal places= 1.2f% , 1.3f%, 1.4f%
plt.pie(expense,labels=activity,autopct="%1.2f%%",colors=["cyan","coral","skyblue","lightgreen","yellow"])
plt.show()
```

# Line plot

```python
import matplotlib.pyplot as plt
import seaborn as sb
import pandas as pd
import numpy as np
```

```python
# sb.lineplot(x=,y=,data=)

df=sb.load_dataset("flights")
df.head(20)
```

```python
sb.lineplot(x="year",y="passengers",data=df,marker="o",markerfacecolor="red")
# plt.xticks(range(1949,1961))
plt.show()
```

**The line and data points show the avg number of passengers in each year.**


**Shaded region**-> Shows the - confidence interval.

It represents the spread of the data.

Wide shade region width(wider)-> more spread.

Less shade region width(narrow)-> less spread



```python
47,51,49,53,50,52  -> numbers are close to each other.
                      Less Spread

50,70,90,51,49,91  -> more spread
```

```python
sb.lineplot(x="year",y="passengers",data=df,marker="o",markerfacecolor="red",errorbar=None)
# Remove the spread(confidence interval) -> errorbar=None
```

By default the data point on the line are the **mean**.

estimator="mean" (default)

--

We can change the aggregation->

estimator="sum", "count","median"

```python
# Total passengers in each year-> sum
sb.lineplot(x="year",y="passengers",data=df,marker="o",markerfacecolor="red",errorbar=None,estimator="sum")
plt.show()
```

```python
# Total passengers, month vise

sb.lineplot(x="year",y="passengers",data=df,hue="month",estimator="sum",marker="*")
plt.show()
```
