```python
'''
1. Create scatter plots for the following data, make sure all the plots appear in the same
plane with x-labels and y-labels evenly spaced.

a. Create scatter plots with between (mpg-hp, mpg-weight), and (hp-mpg,
hp-time-to-60) with respect to each brand.
'''

# --22:00

# scatter plots - they are created for two numeric variable if you want to study the
# relationship between them


fig = plt.subplots(2, 2, figsize = (20, 10))

plt.subplot(2,2,1)
sns.scatterplot(x=df['mpg'], y= df['hp'], data = df, hue = df['brand'])

plt.subplot(2,2,2)
sns.scatterplot(x=df['mpg'], y= df['weightlbs'], data = df, hue = df['brand'])

plt.subplot(2,2,3)
sns.scatterplot(x=df['hp'], y= df['mpg'], data = df, hue = df['brand'])

plt.subplot(2,2,4)
sns.scatterplot(x=df['hp'], y= df['time-to-60'], data = df, hue = df['brand'])

plt.show()
```

```python
'''
2. Create bar plots for the following, make sure all the plots appear in the same plane with
x-labels and y-labels evenly spaced.
a. Create a bar plot that shows the visual representation of hp, mpg, weight,
time-to-60 with respect to the number of cylinders for each of the brands.

hp vs cylinder
mpg vs cylinder
weightlbs vs cylinder
time-to-60 vs cylinder
'''

fig = plt.subplots(2, 2, figsize=(20, 10))

plt.subplot(2, 2, 1)
sns.barplot(x='cylinders', y='hp', data=df, hue='brand'), plt.title("hp vs cylinders"), plt.xlabel("cylinders"), plt.ylabel("hp")

plt.subplot(2, 2, 2)
sns.barplot(x='cylinders', y='mpg', data=df, hue='brand'), plt.title("mpg vs cylinders"), plt.xlabel("cylinders"), plt.ylabel("mpg")

plt.subplot(2, 2, 3)
sns.barplot(x='cylinders', y='weightlbs', data=df, hue='brand'), plt.title("weight vs cylinders"), plt.xlabel("cylinders"), plt.ylabel("weightlbs")

plt.subplot(2, 2, 4)
sns.barplot(x='cylinders', y='time-to-60', data=df, hue='brand' ), plt.title("time-to-60 vs cylinders"), plt.xlabel("cylinders"), plt.ylabel("time-to-60")

plt.tight_layout()
plt.show()
```

```python
'''
3. Create box plots for the following:
a. Create box plots for the columns in the step1 with respect to the number of
cylinders and brand respectively.


hp vs cylinder
mpg vs cylinder
weightlbs vs cylinder
time-to-60 vs cylinder
'''

# 21:28


plt.figure(figsize=(20, 10))

plt.subplot(2,2,1)
sns.boxplot(x='cylinders', y='hp', hue='brand', data = df)
plt.title("HP vs Cylinder")

plt.subplot(2,2,2)
sns.boxplot(x='cylinders', y='mpg', hue='brand', data = df)
plt.title("MPG vs Cylinder")

plt.subplot(2,2,3)
sns.boxplot(x='cylinders', y='weightlbs', hue='brand', data = df)
plt.title("Weight vs Cylinder")

plt.subplot(2,2,4)
sns.boxplot(x='cylinders', y='time-to-60', hue='brand', data = df)
plt.title("Time to 60 vs Cylinder")

plt.tight_layout()

plt.show()
```

```python
'''
4. Create pair plots for the entire data to study various patterns in the data.
a. Create pair plots with respect to brand, number of cylinders, year, etc.

Pair of Area plot and Scatter plot
'''

sns.pairplot(df, hue='brand')
```

```python
'''
6. Create a histogram for the following:
a. Create histograms for hp, mpg, time-to-60, weightlbs, cubic inches, year, with
respect to brand.

'''

plt.figure(figsize=(12,10))


plt.subplot(3,2,1)
sns.histplot(data=df, x="hp", hue="brand")


plt.subplot(3,2,2)
sns.histplot(data=df, x="mpg", hue="brand")


plt.subplot(3,2,3)
sns.histplot(data=df, x="time-to-60", hue="brand")


plt.subplot(3,2,4)
sns.histplot(data=df, x="weightlbs", hue="brand")


plt.subplot(3,2,5)
sns.histplot(data=df, x="cubicinches", hue="brand")


plt.subplot(3,2,6)
sns.histplot(data=df, x="year", hue="brand")


plt.tight_layout()
plt.show()
```
