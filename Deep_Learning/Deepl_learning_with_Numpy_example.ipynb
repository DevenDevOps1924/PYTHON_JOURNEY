```python
import numpy as np
```

```python
choclate = 2
icecream = 5
bottle = 3

total_price = 100
```

```python
learning_rate = 0.01
w1 = 0
w2 = 0
w3 = 0
```

```python
for epoch  in range(20):
  total_price_predicition = w1*choclate +w2*icecream +w3*bottle

  error = (total_price - total_price_predicition)**2

  dew1 = 2*(total_price - total_price_predicition)*(-choclate)
  dew2 = 2*(total_price - total_price_predicition)*(-icecream)
  dew3 = 2*(total_price - total_price_predicition)*(-bottle)



  print("Epoch:", epoch, "Predicted:", round(total_price_predicition, 2), "Error:", round(error, 2),
          "w1:", round(w1, 2), "w2:", round(w2, 2), "w3:", round(w3, 2))

  w1 = w1-learning_rate*dew1
  w2 = w2-learning_rate*dew2
  w3 = w3-learning_rate*dew3

```

```python
x1 = 4
x2 = 5
x3 = 3

y = 100
w1 = 0
w2 = 0
w3 = 0
lr = 0.01
```

```python
import numpy as np
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
for epoch in range(50):
  predict = w1*x1+w2*x2+w3*x3
  predict_2 = sigmoid(predict)
  error = (y-predict_2)**2


  dew1 = 2*(y - predict_2)*(-x1)
  dew2 = 2*(y - predict_2)*(-x2)
  dew3 = 2*(y - predict_2)*(-x3)

  w1 = w1-lr*dew1
  w2 = w2-lr*dew2
  w3 = w3-lr*dew3


  print("Epoch:", epoch, "Predicted:", round(predict_2, 2), "Error:", round(error, 2),
          "w1:", round(w1, 2), "w2:", round(w2, 2), "w3:", round(w3, 2))




```
