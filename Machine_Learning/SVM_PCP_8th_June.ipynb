# **SVM - Support Vector Machines**

Svm is a powerful ML model used for both classification and regression.

It's a supervised ML model.


Suppose we have two classes ( spam vs not spam, positive vs negative, yes no).

We want to find a line / decision boundary which best seperates the classes.

Suppose you work for a bank and you want to classify customers into two groups:

* Customers who are likely to repay a loan
* Customers who are likely to default

Or suppose you work for an email company and want to classify emails into:

* Spam
* Not Spam

Or in healthcare:

* Disease Present
* Disease Absent

In all these cases, our goal is to separate data points into different categories.

Now, there are many algorithms that can perform classification, such as Logistic Regression, Decision Trees, Random Forests, and so on.

So why do we need another algorithm called SVM?

**We seperate the classes using a line or a hyperplane**

There can be a lot of lines that seperate the classes.

But, SVM says:

"I don't want just any line that separates the classes. I want the line that leaves the maximum possible distance from both classes."

In simple words, SVM tries to create the safest possible boundary between classes.

Why?

Because when new unseen data arrives, a larger safety gap usually leads to better predictions and better generalization.

This safety gap is called the Margin.

The main objective of SVM is:

Find the decision boundary that maximizes the margin between the classes.

This is the key idea behind the entire SVM algorithm.

The name Support Vector Machine comes from the fact that only a few critical points, called Support Vectors, determine the position of this boundary.

  Correct classification

# Svm For Regression

While Support Vector Machines are famous for classification, they are equally powerful when adapted for regression tasks. This variation is known as Support Vector Regression (SVR).

The fundamental mindset shift from classification to regression is this:

In Classification, we try to fit a hyperplane that maximizes the margin between two distinct classes (keeping data points out of the margin).

In Regression, we reverse this logic. We try to fit a tube around our line of best fit that encloses as many data points as possible inside the margin.

## Allowing the model to have a margin of error.

There are three major reasons why this design choice is incredibly powerful in practice.

**1. Prevention of Overfitting (Generalization)**

Real-world data is inherently noisy. It contains random fluctuations, measurement errors, and minor anomalies.

If you force a model to penalize every single micro-error (like standard Ordinary Least Squares linear regression does), the model will constantly twitch and bend trying to pass perfectly through every fluctuating data point. It learns the noise rather than the signal.

By setting a margin of error ($\epsilon$), you tell the model: "If a data point is only off by a tiny fraction, ignore it. It’s just noise." This keeps the regression line smooth and stable, allowing it to perform much better on new, unseen data.

**2. Extreme Robustness to Outliers**

In traditional linear regression, a single extreme outlier can pull the entire line of best fit completely out of whack because the algorithm squares the errors ($Loss = \sum e^2$). The further away a point is, the harder it pulls the line.

A. The Inside-Tube Immunity

If an outlier happens to fall inside or right on the boundary of the $\epsilon$-tube, its penalty is exactly zero. The model does not care how far it is from the center line; as long as it is within the tolerance zone, it has zero influence on the model’s coefficients.

B. Linear Penalization (The Cap on Power)


For points that fall outside the tube, SVR does not square the distance. It penalizes them linearly.If a point drifts 20 units away, its penalty is scaled linearly based on your regularization parameter ($C$). Because it isn't squared, an outlier cannot exponentially bully the model into tilting the regression line. The line stays anchored to the dense cluster of normal data points.

**3. Computational Efficiency and Sparsity**

When a model ignores errors inside the tube, the optimization algorithm sets the weights (coefficients) for those internal data points to exactly zero. They are completely dropped from the final prediction equation.

This results in a sparse solution. Instead of needing to store and calculate calculations for all 10,000 training points when making a new prediction, the model only needs to remember the 500 points that became Support Vectors. This drastically reduces memory usage and speeds up prediction times when deploying the model to production.

# Working of SVR( Support Vector Regression)



# Understanding Support Vector Regression (SVR) From Scratch

Support Vector Regression (SVR) adapts the core logic of Support Vector Machines (SVM) to solve regression problems. Instead of trying to keep data points *out* of a margin (like in classification), SVR tries to fit a "tube" around the data that encloses as many points as possible *inside* the margin.

---

## Step 1: Initialize the Linear Equation
SVR starts with the standard equation for a hyperplane (a straight line in 2D space):

$$y = w \cdot x + b$$

* **w**: The weights (or slope) of the features.
* **b**: The bias (or y-intercept).

The goal of SVR is to find the optimal values for $w$ and $b$ based on the training data.

---

## Step 2: Establish the $\epsilon$-Insensitive Tube
Before calculating any errors, SVR introduces a tolerance zone called the **$\epsilon$-insensitive tube** (epsilon tube). You define a value for $\epsilon$, and the model creates two virtual boundary lines around our main prediction line:

* **Upper Boundary:** $w \cdot x + b + \epsilon$
* **Lower Boundary:** $w \cdot x + b - \epsilon$



This creates a corridor of total width $2\epsilon$. Any data point that falls completely inside this corridor is considered a "perfect prediction" and contributes **zero** to the training error calculation.

---

## Step 3: Handle the Outliers (Slack Variables)
Inevitably, some data points will fall outside this tube. SVR measures how far these outlying points are from the tube's boundaries using **Slack Variables**, denoted as $\xi$ (ksi) and $\xi^*$ (ksi-star):

* $\xi_i$: The distance from the *upper* boundary to a data point above the tube.
* $\xi_i^*$: The distance from the *lower* boundary to a data point below the tube.

If a point sits safely inside or exactly on the boundary of the tube, both $\xi_i$ and $\xi_i^*$ are exactly $0$.

---

## Step 4: Solve the Core Optimization Problem
To find the final position of the line, SVR minimizes a balancing mathematical equation called the **Objective Function**:

$$\min_{w, b, \xi, \xi^*} \left( \frac{1}{2} ||w||^2 + C \sum_{i=1}^{n} (\xi_i + \xi_i*) \right)$$

This equation forces the model to balance two opposing goals:

### Part A: Flatness ($\frac{1}{2} ||w||^2$)
By minimizing the norm of the weights ($||w||^2$), SVR forces the weights to be as small as possible. Geometrically, smaller weights mean a **flatter, smoother line**. A flatter line is less sensitive to minor fluctuations and generalizes much better to unseen data.

### Part B: Penalty ($C \sum (\xi_i + \xi_i^*)$)
This part calculates the total distance of all points living outside the tube.

* **The C Parameter** acts as a scale slider. If $C$ is massive, the model prioritizes smashing the slack variables down to zero (making a line that bends aggressively to accommodate outliers). If $C$ is small, the model focuses entirely on making the line flat, letting outliers fall wherever they want.

---

## Step 5: Identify the Support Vectors
Once the optimization algorithm finishes balancing flatness versus error penalties, it locks the line into place.

The algorithm discovers that it doesn't need all the data points to hold the line up. The weights $w$ are completely determined by the data points that sit **on the boundary or outside the tube**.

These specific boundary-defining points are named the **Support Vectors**. Every point safely inside the tube gets a mathematical multiplier of zero—meaning they are completely ignored when drawing the final line.

---

## Summary of the Creation Flow
1. **The model sets up a straight line** with initial weights ($w$) and bias ($b$).
2. **It inflates a corridor** of width $\epsilon$ around that line.
3. **It evaluates the points outside the corridor** and tracks their distances ($\xi, \xi^*$).
4. **The optimizer tilts and shifts the line** until it finds the perfect mathematical sweet spot: the flattest possible line that minimizes the total external distance, weighted by $C$.
5. **The final line relies entirely on the support vectors** to maintain its orientation.

**Finding the optimal value for $\epsilon$ (epsilon)**

is one of the most critical parts of building a Support Vector Regression model, because it directly controls how many data points become Support Vectors and how much noise your model ignores.Unlike hyperparameters like $C$, which are purely mathematical tuning knobs, $\epsilon$ can often be estimated directly from your data. Here is how you find the right $\epsilon$ value for your tube.

**The Statistical Approach**

(Data-Driven)If you don't have deep domain expertise to pick an arbitrary number, you can let the inherent noise of your dataset guide you. A common data-driven heuristic is to tie $\epsilon$ to the Standard Deviation ($\sigma$) or the Mean Absolute Error (MAE) of a baseline model.

A highly regarded research paper ("Selecting Support Vector Hyperparameters" by Cherkassky and Ma) suggests setting $\epsilon$ based on the standard deviation of the noise ($\sigma$) in the target variable $y$:$$\epsilon = 3 \sigma \sqrt{\frac{\ln(n)}{n}}$$

(Where $n$ is the number of training samples, and $\sigma$ is estimated from the residuals of a simple baseline model like a quick Linear Regression or K-Nearest Neighbors).

In simpler terms: Look at the average noise or variance in your target data, and set $\epsilon$ to a small fraction of that variance so the tube acts as a natural noise filter.

*$\epsilon$ is too small* (e.g., 0): The tube is a thin wire. Almost every single data point falls outside the tube and becomes a Support Vector. The model tries to fit every point, making training slow, memory usage high, and risking overfitting.

*$\epsilon$ is too large:* The tube is massive. Every data point fits easily inside the tube. Because no points are outside, the model has zero penalties to minimize, resulting in a flat, useless line that underfits the data.

# Split data into X and Y

```python
# target col= "letter"

x= df.drop("letter",axis=1)
y= df["letter"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

```python
scaler= StandardScaler()

x_train_scale = scaler.fit_transform(x_train)
x_test_scale  = scaler.transform(x_test)

x_train_scale
```

## Baseline SVM



```python
# classification -> SVC(support vector classifier)

from sklearn.svm import SVC
```

```python
svm = SVC()

# Train

svm.fit(
    x_train_scale, # Training Input
    y_train        # Training Output
)

# Predictions
y_pred= svm.predict(
    x_test_scale
)
```

```python
print("Accuracy=",accuracy_score(y_test,y_pred)*100,"%")
```

### Trying the different types of kernel

```python
kernels=["sigmoid","linear","poly","rbf"]

result=[]

for i in kernels:
  svm = SVC(
      kernel=i
            )

  # Train

  svm.fit(
    x_train_scale, # Training Input
    y_train        # Training Output
  )

  # Predictions
  y_pred= svm.predict(
    x_test_scale
  )
  acc = accuracy_score(y_test, y_pred)
  result.append([i,acc])


```

```python
result_df = pd.DataFrame(
    result,
    columns=["Kernel","Accuracy"]
)

result_df
```

## Grid Search CV

```python
param_grid={
    "kernel":["poly","rbf"],
    "C":[0.1, 0.2, 1 ,5],       # hinge loss panelty
    "gamma":[0.01,0.1,0.2],  # rbf
    "degree":[1, 2, 3]
}
```

```python
grid= GridSearchCV(
    # model / estimator= SVC()
    estimator=SVC(),
    # Parameter grid -> param_grid
    param_grid= param_grid,
    # cv - cross validate
    cv=3,
    # n_jobs -> number of cpu cores to use ( -1 use all)
    n_jobs=-1
)

grid.fit(
    x_train_scale,
    y_train
)
```

```python
print("Best Parameters: ",grid.best_params_)
print("Best accuracy: ",grid.best_score_)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,      # 40% test data
    random_state=42,    # for reproducibility
    stratify=y          # maintain class proportions
)
```

```python
# 2d-
# Decision boundary
# w1 x1 + w2 x2 + b = 0
# Upper Boundary
# w1 x1 + w2 x2 + b = +1
# Lower Boundary
# w1 x1 + w2 x2 + b = -1

w = svm.coef_[0]   # w1, w2
b = svm.intercept_[0] # b

print("Weights:", w)
print("Intercept:", b)
```

```python
from sklearn.inspection import DecisionBoundaryDisplay

fig, ax = plt.subplots(figsize=(8, 6))
DecisionBoundaryDisplay.from_estimator(
    svm,
    X_test,  # Original dataset used to determine plot range

    response_method="decision_function",

    # When SVM trains, it doesn't just output a hard 0 or 1.
    # It gives every point a score based on how far it is from the border.
    #That score is what the code calls decision_function.
    # --response_method="decision_function" in the code,
    # --you are telling the plotter:
    # "Bro, check the score for every single pixel on this grid."
    # Then, by adding levels=[-1, 0, 1]
    # you are saying: "Draw a line ONLY where the score hits exactly -1, 0, and 1.


    plot_method="contour",
    # decision_function gives every spot on the graph a "score" (or height)
    # contour is the tool that draws lines connecting the places that have the exact same score.

    levels=[-1, 0, 1],  # -1 = negative margin, 0 = decision boundary, 1 = positive margin
    colors="black",
    linestyles=["--", "-", "--"],  # Dashed for margins, solid for boundary
    linewidths=[1, 2, 1],
    ax=ax, #"Draw this decision boundary inside the specific plot window I already created."
)


plt.scatter(X_train[:, 0], X_train[:, 1],
            c=y_train, cmap="bwr",
            edgecolors="k", label="Train")

plt.scatter(X_test[:, 0], X_test[:, 1],
            c=y_test, cmap="bwr",
            marker="*",
            edgecolors="k", label="Test")

plt.legend()
plt.show()
```
