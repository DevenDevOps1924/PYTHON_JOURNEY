```python
# drop the extra column--- CustomerID
data.drop(columns=['customerID'],inplace=True)
```

```python
data.info()
```

```python
# most of the columns are having data type -------object
# convert categorical column data to numeric data ------- label encoding
# 10:11 p.m
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()

# create a list  with all the object data type column  so that  you can convert those columns into numbers
col_list = []
for i in data.columns:
  if((data[i].dtype=='object') & (i!='Churn')):
    col_list.append(i)

col_list
```

```python
for i in col_list:
  data[i] = le.fit_transform(data[i])
```

```python
data
```

```python
data.info()
```

```python
# divide the data into x and y
x = data.iloc[:,:-1] # independent features
y = data['Churn'] # dependent features
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state =42 , stratify = y)
```

```python
# stratify = y -> it will equally divide the values of dependent column between training and testing
```

```python
from random import Random
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators=100) # number of decision tree you want to use
```

```python
model.fit(x_train,y_train)
```

```python
ans = model.predict(x_test)
```

```python
ans
```

```python
from sklearn.metrics import* # import* we can use now all the evaluation metric
# from sklearn.metrics import accuracy_score , precision_score
```

```python
accuracy_score(ans,y_test)
```
