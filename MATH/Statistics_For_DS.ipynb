```python
#Outliers --->An outlier is a datapoint that is very different from the  rest of the data.
#--->Extreme high or extreme low values present in our data is know as outliers
```

```python
a1=[18,25,30,45,67,78,89,93,250,76,99,271]
```

```python
#How to check outliers
import matplotlib.pyplot as plt
plt.boxplot(a1)
plt.show()
```

```python
#IQR ---Interquatile range
#HOW to Remove the outliers
#1. Find q1(25th percentile) and q3(75th percentile)
#2. Iqr = Q3-Q1
#3. Lower bound(Lb)  ---> Q1 -1.5 *IQR
#4 . Upper bound(UB)----> Q3 +1.5 *IQR


# ---->Any values outside these bounds is an outlier
```

image.png

```python
a1=[18,25,30,45,67,78,89,93,250,76,99,271]
#Step 1
q1=np.percentile(a1,25)
q3=np.percentile(a1,75)
print(q1)
print(q3)
```

```python
#Step 2
IQR=q3-q1
print(IQR)
```

```python
#Step 3
LB=q1-1.5*IQR
UB=q3+1.5*IQR
```

```python
clean_a1=[]
for i in a1:
  if i<LB or i>UB:
    print("The outlier in the data is",i," and it haas been detected")
  else:
    clean_a1.append(i)
```

```python
def outlier_remover(df):
  for col in df.columns:  #for each and every col
    if(df[col].dtype!="object"):
      Q1=df[col].quantile(0.25)
      Q3=df[col].quantile(0.75)

      IQR = Q3-Q1

      LB=Q1-1.5*IQR
      UB=Q3+1.5*IQR

      df=df[(df[col]>=LB)&(df[col]<=UB)]

```

```python
#Given that time is Dinner , What is the probability of a day being sunday?
fav=len(df2[(df2['time']=='Dinner') & (df2['day']=='Sun')])
total=len(df2[df2['time']=='Dinner'])
print(fav/total)
```

```python
#Given that time is Dinner ,What is the probability of a day being saturday or friday?
fav = len(df2[((df2['day'] == 'Sat') | (df2['day'] == 'Fri')) & (df2['time'] == 'Dinner')])
total = len(df2[df2['time'] == 'Dinner'])

print((fav / total)*100)
```
