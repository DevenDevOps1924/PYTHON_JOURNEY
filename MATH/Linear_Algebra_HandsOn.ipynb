## **Property 1: Transpose of a Transposed Matrix is the orginal matrix**

```python
final_matrix = a_transpose.T
print("Transpose of transposed matrix:\n", final_matrix)
```

```python
if np.array_equal(a,final_matrix):
  print("Yes the property is true")
else:
  print("The matrices are different")
```

##**Property 2:(Additive Property):(A+B).T=A.T+B.T**

```python
if np.array_equal((m1+m2).T,m1.T+m2.T):
  print("proved")
else:
  print("not True")
```

# **Property 3:(Multiplicative Property):(AXB).T=B.TxA.T**

```python
A= np.random.randint(10,size=(3,3))
B= np.random.randint(10,size=(3,3))
if(np.array_equal((A@B).T,(B.T)@(A.T))):
  print("Hence proved")
else:
  print("Not proved")
```

5. Create a 3*3 matrix and find the determinant of that matrix.

```python
m1
```

```python
np.linalg.det(m1)
```

- if the determinant is equal to zero than the matrix is singular
- if the determinant is not equal to zero than the matrix is invertible

```python
inv =np.linalg.inv(m1)
```

##if we take dot product of the original matrix and its' inverse it going to equal to identity matrix

```python
result = np.dot(m1, inv)
result
```

```python
np.eye(3)
```

np.allclose() it checks if two matrices are mathematically equal or not, even if there are few floating point errors it will ignore them

```python
if np.allclose(result, np.eye(3)):
  print("Yes our assumption is true")
else:
  print("No our assumption is false")
```
